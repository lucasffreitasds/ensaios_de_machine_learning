```python
#seleciono as variáveis que acho que está relacionada a variável resposta, que é saldo_atual
features = ['idade', 'divida_atual', 'renda_anual',
       'valor_em_investimentos', 'taxa_utilizacao_credito', 'num_emprestimos',
       'num_contas_bancarias', 'num_cartoes_credito', 'dias_atraso_dt_venc',
       'num_pgtos_atrasados', 'num_consultas_credito', 'taxa_juros']

#variável resposta, o que quero estudar
label = ['saldo_atual']
```

```python
## 4.1 R squared
r2_squared = np.round(100*mt.r2_score(y_train, y_pred), 2) 
print('R2 square: {}%'.format(r2_squared)) 

## 4.2 MSE
mse = np.round( mt.mean_squared_error(y_train, y_pred), 2) 
print('A cada previsão, o erro médio é de: U${}'.format(mse))


# 5.0 Conclusão
print('{}% da variação da variável alvo y é reduzida, levando em consideração o preditor'.format(r2_squared))
print('{}% da variação da variável alvo y é “explicada pela variação do preditor x'.format(r2_squared))
```

### Utilizando a outra métrica, o RMSE 

```python
## 4.3 RMSE
rmse = np.sqrt(mse)
print('A cada previsão, o erro médio é de: U${}'.format(rmse))
```

# Testando Métricas com Novos Outliers

### Estrutura da Ideia

1. criar uma cópia do dataframe x_train;
2. selecionar porcentagens (0%, 2%, 5%, 10%, 20% e 30%) das linhas a serem modificadas para outliers. Ou seja, terá 6 resultados diferentes, com níveis de outliers cada vez maiores;
3. Alterar algumas features dessas linhas para virar outliers;
4. treinar um novo modelo para cada variação de linhas modificadas;
5. prever o 'saldo_atual' que é o predict, para cada variação de linhas modificadas;
6. calcular o R², MSE e RMSE para cada variação de linhas modificadas;
7. Guardar os resultados em uma tabela;
8. Comparar e analisar os resultados entre as alterações.

### 1) criando variável que guarda porcentagens e features/colunas alteradas

```python
#proporções de outliers que serão testadas
proporcoes_outliers = [0, 0.02, 0.05, 0.10, 0.20, 0.30]

#colunas que sofrerão alterações, virando outliers
colunas_outliers = [
    'divida_atual',
    'renda_anual',
    'valor_em_investimentos'
]
```

### 2) criando uma ordem aleatória das linhas a serem alteradas

```python
#criando uma varíavel para guardar a aleatoriedade usando biblioteca Numpy e a função armazenada
#dentro dela, chamada random.default_rng que é considerado uma função de fábrica, 
#que serve para criar um gerador de números aleatórios (um objeto da classe Generator).
#42 é um número aleatório que vai padronizar como as linhas serão sortidas

#cria uma ferramenta do NumPy capaz de fazer sorteios, embaralhar e gerar números aleatórios.
random_number_generator = np.random.default_rng(42)

#agora utiliza a ferramenta criada acima e pega os indices de x_train e 
#coloca de fato em ordem aleatória e embaralhada
indices_aleatorios = random_number_generator.permutation(x_train.index)
```

### 3) treinando os modelos

```python
#crio uma variável chamada resultado para guardar os resultados
resultados = []

#faço um looping for agora para treinar todas variaçoes dos modelos
#para cada numero de % faça as seguintes operações
for i in proporcoes_outliers:

    #cria uma copia para preservar o x_train original
    x_train_modificado = x_train.copy()

    #calcula quantas linhas serão alteradas. i é a lista de proporções de outliers
    #int(passa a resposta inteira) e 
    #len(pega o tamanho total de linhas). Nesse caso retorna 9500 linhas
    #do total da amostra, quantas delas virarão outliers. Ex: 9500 * 0.05 (i [2])
    #ex: quantidade de linhas que serão modificadas = 475 (quantidade_outliers)
    quantidade_outliers = int(
        len(x_train_modificado) * i
    )

    #seleciona os indices que serão modificados
    #do total dessas amostragem, vai pegar do primeiro até o ultimo dela
    #exemplo, numero de 9500, a quantidade de outliers foi de 5% (i = 0.05)
    #então vai pegar as posições de 0-474 (475 primeiros numeros com index aleatorios)
    indices_outliers = indices_aleatorios[:quantidade_outliers]

    #multiplica por 10 os valores das colunas selecionadas
    #selecionou as linhas e as colunas
    #*= multiplica todos os valores por 10 e guarda no mesmo lugar aqui
    x_train_modificado.loc[indices_outliers, colunas_outliers]*=10

    #define o modelo usado
    modelo_outliers = LinearRegression()

    #treina o modelo com os dados modificados
    modelo_outliers.fit(
        x_train_modificado,
        y_train
    )

    #faz as previsões
    y_pred_outliers = modelo_outliers.predict(x_train_modificado)

    #calcula as metricas
    r2_outliers = mt.r2_score(
        y_train,
        y_pred_outliers
    )

    mse_outliers = mt.mean_squared_error(
        y_train,
        y_pred_outliers
    )

    rmse_outliers = np.sqrt(mse_outliers)

    #guarda o rsultado desta repetição
    resultados.append({
        'outliers_percentual': i*100,
        'quantidade_outliers': quantidade_outliers,
        'r2_outliers': r2_outliers*100,
        'mse_outliers': mse_outliers,
        'rmse_outliers': rmse_outliers
    })
```

### 4) Criando a Tabela e arredondando os resultados

```python
df_resultados_outliers = pd.DataFrame(resultados)

df_resultados_outliers = df_resultados_outliers.round({
    'outliers_percentual': 0,
    'r2_outliers': 2,
    'mse_outliers': 2,
    'rmse_outliers': 2
})

df_resultados_outliers
```
