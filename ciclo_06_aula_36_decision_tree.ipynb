# Imports

```python
import cv2
import pydotplus
import matplotlib.pyplot as plt

from six import StringIO
from sklearn import datasets as dt
from sklearn import tree as tr
from IPython.display import Image
```

# Criando os dados

```python
iris = dt.load_iris()
x = iris.data[:, 2:]
y = iris.target
```

# Treinando Modelo

```python
tree_clf = tr.DecisionTreeClassifier(max_depth=4)
tree_clf.fit(x, y)
```

# Export Draw

```python
dot_data = StringIO()
tr.export_graphviz(
    tree_clf,
    out_file='tree.dot',
    feature_names=iris.feature_names[2:],
    class_names=iris.target_names,
    rounded=True,
    filled=True
)
```

# Convertendo arquivo .dot para .png
