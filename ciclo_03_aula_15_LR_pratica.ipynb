```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression
```

# Load Dataset

```python
df = pd.read_csv('/home/lucas/projetos/exercicios_ds_ml/dataset/train.csv')
```

```python
#seleciono as variáveis que acho que está relacionada a variável resposta, que é saldo_atual
features = ['idade', 'divida_atual', 'renda_anual',
       'valor_em_investimentos', 'taxa_utilizacao_credito', 'num_emprestimos',
       'num_contas_bancarias', 'num_cartoes_credito', 'dias_atraso_dt_venc',
       'num_pgtos_atrasados', 'num_consultas_credito', 'taxa_juros']

#variável resposta, o que quero estudar
label = ['saldo_atual']
```

```python
#crio x_train e y_train com todas linhas e suas colunas respectivas
x_train = df.loc[:, features]
y_train = df.loc[:, label]
```

# Model Training

```python
#model definition
lr_model = LinearRegression()

#model fit (ajuste, treinamento)
lr_model.fit(x_train, y_train)

#previsão
y_pred = lr_model.predict(x_train)
```

```python
#criei uma nova tabela, chamada df1, com todas linhas relacionadas a coluna id cliente e saldo atual(y)
df1 = df.loc[:, ['id_cliente', 'saldo_atual']]

#criei uma nova coluna, de predição, que produzimos os dados dela na previsão, y_pred
df1['predicted'] = y_pred

df1.head()
```
