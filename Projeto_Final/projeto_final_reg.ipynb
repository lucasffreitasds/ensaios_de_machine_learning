# Linear Regression - Lasso e Ridge

```python
#achando os melhores parâmetros
alpha_range = [0.001, 0.01, 0.1, 1, 10, 100]
max_iter_range = [100, 500, 1000, 5000]

# ---- Lasso ----
resultados_lasso = []

for a in alpha_range:
    for max_iter in max_iter_range:
        model_lasso = Lasso(alpha=a, max_iter=max_iter, random_state=0)
        model_lasso.fit(x_train, y_train)

        y_pred = model_lasso.predict(x_val)

        resultados_lasso.append({
            'alpha': a,
            'max_iter': max_iter,
            'R2': mt.r2_score(y_val, y_pred),
            'MSE': mt.mean_squared_error(y_val, y_pred),
            'RMSE': np.sqrt(mt.mean_squared_error(y_val, y_pred)),
            'MAE': mt.mean_absolute_error(y_val, y_pred),
            'MAPE': mt.mean_absolute_percentage_error(y_val, y_pred)
        })

tabela_lasso = pd.DataFrame(resultados_lasso)

# ---- Ridge ----
resultados_ridge = []

for a in alpha_range:
    for max_iter in max_iter_range:
        model_ridge = Ridge(alpha=a, max_iter=max_iter, random_state=0)
        model_ridge.fit(x_train, y_train)

        y_pred = model_ridge.predict(x_val)

        resultados_ridge.append({
            'alpha': a,
            'max_iter': max_iter,
            'R2': mt.r2_score(y_val, y_pred),
            'MSE': mt.mean_squared_error(y_val, y_pred),
            'RMSE': np.sqrt(mt.mean_squared_error(y_val, y_pred)),
            'MAE': mt.mean_absolute_error(y_val, y_pred),
            'MAPE': mt.mean_absolute_percentage_error(y_val, y_pred)
        })

tabela_ridge = pd.DataFrame(resultados_ridge)
```

# Linear Regression - Elastic Net

Parâmetros: aplha, l1_ratio e max_iter

### LR - Val

```python
#encontrando os melhores parametros

alpha_range = [0.001, 0.01, 0.1, 1, 10, 100]
l1_ratio_range = [0.1, 0.3, 0.5, 0.7, 0.9]
max_iter_range = [100, 500, 1000, 5000]

# ---- Elastic Net ----
resultados_elastic = []

for a in alpha_range:
    for l1 in l1_ratio_range:
        for max_iter in max_iter_range:
            model_elastic = ElasticNet(alpha=a, l1_ratio=l1, max_iter=max_iter, random_state=0)
            model_elastic.fit(x_train, y_train)

            y_pred = model_elastic.predict(x_val)

            resultados_elastic.append({
                'alpha': a,
                'l1_ratio': l1,
                'max_iter': max_iter,
                'R2': mt.r2_score(y_val, y_pred),
                'MSE': mt.mean_squared_error(y_val, y_pred),
                'RMSE': np.sqrt(mt.mean_squared_error(y_val, y_pred)),
                'MAE': mt.mean_absolute_error(y_val, y_pred),
                'MAPE': mt.mean_absolute_percentage_error(y_val, y_pred)
            })

tabela_elastic = pd.DataFrame(resultados_elastic)
```

# Polinomial Regression

```python
#treinando o melhor parâmetro

degree_range = np.arange(1, 6, 1)   # graus baixos — cresce muito rápido em número de colunas

resultados_poly = []

for d in degree_range:
    poly = PolynomialFeatures(degree=d)
    x_train_poly = poly.fit_transform(x_train)
    x_val_poly = poly.transform(x_val)

    model_poly = LinearRegression()
    model_poly.fit(x_train_poly, y_train)

    y_pred = model_poly.predict(x_val_poly)

    resultados_poly.append({
        'degree': d,
        'R2': mt.r2_score(y_val, y_pred),
        'MSE': mt.mean_squared_error(y_val, y_pred),
        'RMSE': np.sqrt(mt.mean_squared_error(y_val, y_pred)),
        'MAE': mt.mean_absolute_error(y_val, y_pred),
        'MAPE': mt.mean_absolute_percentage_error(y_val, y_pred)
    })

tabela_poly = pd.DataFrame(resultados_poly)
```
